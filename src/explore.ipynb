```python
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    if len(outliers) > 0:
        print(f"\nColonne : {col}")
        print(outliers[col])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Mental_Health_Risk')

plt.title("Distribution de la variable cible")
plt.xlabel("Classes")
plt.ylabel("Nombre d'observations")

plt.show()
```

```python
target = 'Mental_Health_Risk'

df.groupby(target).mean(numeric_only=True).T
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x=target,
    y='Sleep_Hours'
)

plt.title('Relation entre les heures de sommeil et le risque')
plt.xlabel('Niveau de risque')
plt.ylabel('Heures de sommeil')

plt.show()
```

```python
df[target].value_counts(normalize=True) * 100
```

```python
plt.figure(figsize=(14, 10))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Matrice de corrélation")

plt.show()
```

```python
numeric_cols = df.select_dtypes(include='number').columns


for col in numeric_cols:

    plt.figure(figsize=(8, 5))

    sns.boxplot(data=df, x=target, y=col)

    plt.title(f'{col} selon le niveau de risque')
    plt.xlabel('Niveau de risque')
    plt.ylabel(col)

    plt.show()
```
